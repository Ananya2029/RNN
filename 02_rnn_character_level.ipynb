```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense
```

```python
# Create character sequence: A to Z
sequence = list("ABCDEFGHIJKLMNOPQRSTUVWXYZ")

print(sequence)
```

```python
# Define window size: 3 characters → 1 output
window_size = 3

# X = input characters
# y = next character
X = []
y = []
```

```python
# Create input-output pairs
for i in range(len(sequence) - window_size):
    X.append(sequence[i:i + window_size])
    y.append(sequence[i + window_size])

print("First 5 input examples:")
print(X[:5])

print("First 5 answers:")
print(y[:5])
```

```python
# Create character-to-number mapping
chars = sorted(set(sequence))

char_to_number = {
    char: i for i, char in enumerate(chars)
}

number_to_char = {
    i: char for char, i in char_to_number.items()
}

print("Character to number:")
print(char_to_number)
```

```python
# Convert X characters to numbers
X = np.array([
    [char_to_number[char] for char in row]
    for row in X
])

# Convert y characters to numbers
y = np.array([
    char_to_number[char]
    for char in y
])

print("X:")
print(X[:5])

print("y:")
print(y[:5])
```

```python
# Reshape X to: samples, timesteps, features
X = X.reshape((X.shape[0], X.shape[1], 1))

print("New X shape:", X.shape)
```

```python
model = Sequential()

model.add(
    SimpleRNN(
        50,
        activation='relu',
        input_shape=(window_size, 1)
    )
)

model.add(
    Dense(len(chars), activation='softmax')
)

model.summary()
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
model.fit(
    X,
    y,
    epochs=500,
    verbose=2
)

print("Training complete.")
```

```python
# Give a new input
test_input = ['A', 'B', 'C']

# Convert characters to numbers
test_input = np.array([
    char_to_number[char]
    for char in test_input
])

# RNN expects:
# samples, timesteps, features
test_input = test_input.reshape(
    (1, window_size, 1)
)

# Predict
predicted = model.predict(
    test_input,
    verbose=0
)

# Get character with highest probability
predicted_index = np.argmax(predicted[0])

# Convert number back to character
predicted_char = number_to_char[predicted_index]

print("Input:", ['A', 'B', 'C'])
print("Predicted next character:", predicted_char)
```
